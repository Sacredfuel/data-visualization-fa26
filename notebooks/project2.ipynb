# Final Project

This is the final project for the course. You can work in teams of 1–2.

Your goal: take a dataset from [Kaggle](https://www.kaggle.com/) and build an **animated visualization in D3 on Observable** that reveals an insight about the data that animation shows **better** than a static chart can. To prove it, you'll also build the best **static** version you can in **Altair**, and argue why the animation wins.

This project brings together both halves of the course: Altair for the static chart, D3 and Observable for the animation.

The project is worth 100 points total. Here is the points breakdown for each component:

## The Big Idea: Why Animate?

In Lecture 18 we saw that animation is often *worse* than static charts for careful analysis: it's hard to compare a frame you're looking at with one you saw ten seconds ago. So an animation has to earn its place.

Animation is genuinely good at a few things. Your insight should come from one of them:

- **Tracking individuals through change:** one item moving against the crowd (like China's famine in the Gapminder chart).
- **Rank swaps and overtakes:** who passed whom, and when (like the Big Mac bar chart race).
- **Order and timing:** what happened first, what spread where, what happened at the same time.
- **Rate of change:** speeding up, slowing down, sudden jumps.
- **Connecting two views:** following each item from one layout to another (like the NYC map turning into a scatter plot).

**The test:** put your best static Altair chart next to your D3 animation. Could someone spot your insight in the static chart just as easily? If yes, it's not the right insight (or not the right dataset). If no, explain why not. 

### Project Proposal

Due Satuday, 28 November 2026 EoD

The proposal is split into 4 sections:

#### Overview

Think of the overview section as the equivalent of an abstract in a research paper, or an elevator pitch for the project. These questions will help you frame your thoughts if you ever have to describe your project in an interview:

- **Title**: should capture the topic/theme of your project.
- **Team**: names of all team members.
- **Objective**: In 1 to 2 sentences, describe what you are hoping to show, in simple, non-technical English.
- **Importance**: In 1 to 2 sentences, describe why this project has personal significance to you.

#### The Insight

This is the most important part of the proposal. It's fine if your final insight changes once you see the data, but start with a hypothesis.

- **Hypothesis**: In 1 to 2 sentences, what do you expect the animation to reveal?
- **Why animation**: Which of the strengths listed above does it rely on (tracking individuals, rank swaps, order and timing, rate of change, connecting two views)?
- **Why not static**: In 2 to 3 sentences, why do you expect a static chart to struggle to show this?

#### Prior Work

What will you be basing your animation on? Find an animated visualization that interests you and that suits your data. Good places to look:

- The [D3 gallery on Observable](https://observablehq.com/@d3/gallery) (its examples run directly in Observable, which will save you time)
- The [D3 Graph Gallery](https://d3-graph-gallery.com/)
- Our lectures: the Gapminder bubble chart and bar chart race (Lecture 18), morphing projections, the flight globe, and the map-to-scatter morph (Lecture 19)

It should not be a simple line or bar chart that just plays over time. Make sure to cite the work you're building on, and in 1 to 2 sentences say **what you'll change or add**.

#### Data

With your animation in mind, find a dataset on [Kaggle](https://www.kaggle.com/datasets) that would produce interesting results.

Make sure to comment on the following:

- **Data Source**: The Kaggle URL(s) for your dataset(s), and the dataset's license. All data must be publicly available.
- **Data Volume**: How many rows and columns? If you are joining multiple datasets, how many rows and columns remain after merging?
- **Data Richness**: What type of data is in your dataset? A general overview is fine; you don't need to describe every column. If the dataset has a data dictionary, link to it.
- **Animation Driver**: Which column (or columns) will drive the animation? Usually this is a date or year, but it can also be different states to move between, like two layouts or different groupings. **If your data has nothing to animate between, pick a different dataset.**
- **Size Plan**: Observable runs in the browser, so very large files won't load well. If your dataset is large, explain how you'll filter or aggregate it (for example, monthly totals instead of individual transactions, or the top 50 items instead of all 10,000).

#### Requirements for the D3 animation (Observable)

- Built with **D3** in an Observable notebook. You may use Observable's Plot function for supporting charts, but the main animation must be D3 svg.
- Uses a **key function** in the data join so each mark represents the same item throughout the animation (object constancy, Lecture 18).
- Animates with **transitions**, with sensible timing and easing (for example, linear easing when time drives the animation).
- Gives the viewer **control**: at least a play/pause button or a scrubber. Viewers should never be stuck waiting for an animation.
- Includes at least one more **interaction**, such as tooltips, highlighting an item, a dropdown that changes what's shown, or zoom/pan.

#### Requirements for the static chart (Altair)

- Uses the **same data** and tries to show the **same insight** as your animation.
- Is a **genuine best effort**: use what you learned in the first half of the course, such as faceting into small multiples, layering, color, and highlighting. A deliberately weak static chart that exists only to lose won't earn credit; your argument is only convincing if the static version is good.
- You can include more than one static attempt if you tried several approaches. That's encouraged.

### Project Report

Due Friday, 18 December 2026

Your report is the argument for your animation. Write it as `REPORT.md` in your repository (or as a PDF linked from your README), **1,200–2,000 words**, with these sections:

1. **Question and Data**: What were you curious about? Where does the data come from, and what did you do to prepare it?
2. **The Insight**: State your insight in one or two sentences. Be specific: name the items, the time period, and the numbers.
3. **Static vs. Animated**: Show your Altair chart(s) and screenshots or a short GIF of your animation side by side. Argue why the insight is easier to see, or only noticeable, in the animated version. Be honest: if the static chart does some things better, say so.
4. **Design Decisions**: How did you encode the data (what's on x, y, color, size)? What did you use as the key, and why? How did you choose the animation's speed and easing? What interaction did you add, and why?
5. **Limitations**: What could mislead a viewer? For example, transitions between data points invent the in-between values, and fast-moving scenes can make people miss changes. What would you do with more time?
6. **References**: Cite your dataset, the prior work you built on, and any other sources.

### Tips

- **Pick your dataset with the animation in mind.** Kaggle datasets with a time column work well: sports seasons, music charts over the years, video game sales by year, Olympic results, climate and energy data, prices over time. Before committing, quickly plot a few items over time in Altair. If nothing interesting changes, the animation won't be interesting either.
- **Smaller is usually better.** 10–50 items moving on screen is easy to follow. 5,000 is noise. Filter to the items that tell your story, or let the viewer choose.
- **Start from working code.** Adapt the lecture examples or a gallery example to your data first, then customize. Don't write the animation from a blank cell.
- **Get the static chart done early.** Building the best static version often shows you what your insight really is, and what the animation needs to do better.
